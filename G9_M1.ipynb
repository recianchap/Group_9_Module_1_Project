```python
########## Data Load Module

import pandas as pd

# Show all rows
pd.set_option("display.max_rows", None)

# Show all columns
pd.set_option("display.max_columns", None)

# Show full text content without clipping wide strings
pd.set_option("display.max_colwidth", None)   

df = pd.read_csv('SGJobData.csv')  

# Get row and column counts
rows, cols = df.shape
print(f"Successfully loaded {rows:,} records across {cols} columns.")
```

```python
########## Data Analysis Module

#Data Overview
def print_structured_info(df):
    print("=" * 45)
    print("           DATASET OVERVIEW")
    print("=" * 45)
    print(f" Total Rows    : {len(df):,}")
    print(f" Total Columns : {len(df.columns):,}")
    print(
        f" Memory Usage  : {df.memory_usage(deep=True).sum() / (1024 ** 2):.2f} MB"
    )
    print("=" * 45)
    print("\n--- Column Breakdown ---")

    summary = pd.DataFrame(
        {
            "Data Type": df.dtypes,
            "Non-Null": df.notnull().sum(),
            "Null Count": df.isnull().sum(),
            "Missing (%)": (df.isnull().mean() * 100).round(2),
            "Unique Values": df.nunique(),
        }
    )
    print(summary.to_string())

# Display the structured information
print_structured_info(df)
```
